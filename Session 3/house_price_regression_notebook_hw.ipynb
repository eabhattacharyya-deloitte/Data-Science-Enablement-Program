## Step 4 — Data Preprocessing

**What to do:**  
Prepare the data so it is suitable for linear regression modelling.

1. **Separate features and target** — define X (feature matrix) and y (target vector)
2. **Train-Test Split** — split data 80/20; use a fixed `random_state` for reproducibility. Training set is used to fit the model; test set is held out for unbiased evaluation
3. **Feature Scaling** — apply `StandardScaler` (zero mean, unit variance) to the training set, then apply the *same* fitted scaler to the test set. This prevents data leakage and makes coefficients directly comparable

## Step 6 — Assumptions of Linear Regression

**Why this matters:**  
OLS estimates are only **BLUE** (Best Linear Unbiased Estimators) when all five Gauss-Markov assumptions hold. Violating them leads to biased coefficients, incorrect standard errors, and unreliable p-values.

### The 5 Assumptions to Check:

| # | Assumption | Consequence if Violated | Diagnostic |
|---|-----------|------------------------|------------|
| 1 | **Linearity** — relationship between X and y is linear | Biased estimates | Residuals vs Fitted plot |
| 2 | **Independence** — residuals are not correlated with each other | Underestimated std errors | Durbin-Watson test |
| 3 | **Homoscedasticity** — constant residual variance | Inefficient estimates, wrong CIs | Scale-Location plot |
| 4 | **Normality** — residuals are normally distributed | Invalid hypothesis tests | Q-Q plot + Shapiro-Wilk |
| 5 | **No multicollinearity** — features are not highly correlated | Unstable, uninterpretable coefficients | VIF (Variance Inflation Factor) |

## Step 7 — Model Evaluation Metrics

**What to do:**  
Quantify how well the model predicts house prices using five standard regression metrics. Always report **both training and test** metrics to detect overfitting.

### Metric Definitions

| Metric | Formula | Interpretation |
|--------|---------|----------------|
| **R2** | 1 - SS_res/SS_tot | Proportion of variance in price explained by the model. Range 0–1 |
| **Adjusted R2** | 1 - (1-R2)(n-1)/(n-p-1) | R2 penalised for adding unhelpful features. Better for model comparison |
| **MAE** | mean(|y - yhat|) | Average absolute prediction error in $1,000s. Robust to outliers |
| **MSE** | mean((y - yhat)^2) | Mean squared error. Heavily penalises large mistakes |
| **RMSE** | sqrt(MSE) | Root mean squared error — same units as target ($1,000s). Most reported metric |

**Key diagnostic:** If Train R2 >> Test R2, the model is overfitting.

## Step 8 — Full OLS Statistical Summary (statsmodels)

**What to do:**  
Use `statsmodels.OLS` to get the full statistical output including:
- **Coefficient estimates** and their **standard errors**
- **t-statistics** and **p-values** — to test whether each coefficient is statistically significant (H0: coef = 0)
- **95% Confidence Intervals** — range of plausible values for each coefficient
- **F-statistic** — tests whether the model as a whole is significant (H0: all coefficients = 0)
- **AIC / BIC** — information criteria useful for comparing multiple models

**Decision rule:** If p-value < 0.05, that feature is statistically significant at the 5% level.
